#### 1. Find duplicate elements in an array

```python
def find_duplicates(nums):
    seen = set()
    duplicates = set()

    for num in nums:
        if num in seen:
            duplicates.add(num)
        else:
            seen.add(num)

    return duplicates

print(find_duplicates([1, 2, 3, 2, 4, 5, 1, 5]))
```

#### 2. Find the first non-repeating character in a string

```python
def find_non_repeating_char(string):
    count = {}

    for char in string:
        count[char] = count.get(char, 0) + 1

    for char in string:
        if count[char] == 1:
            return char

    return None

print(find_non_repeating_char("swiss"))
```

#### 3. Check if two strings are anagrams

```python
# Method 1: Using sorting

def anagram(s1, s2):
    return sorted(s1) == sorted(s2)

print(anagram("listen", "silent"))
```

```python
# Method 2: Custom Code

def is_anagram(s1, s2):

    if len(s1) != len(s2):
        return False

    count = {}

    for ch in s1:
        count[ch] = count.get(ch, 0) + 1

    for ch in s2:
        if ch not in count:
            return False

        count[ch] -= 1

        if count[ch] < 0:
            return False

    return True

print(is_anagram("listen", "silent"))
```

#### 5. Common Elements Between Two Lists

```python
# Use: Finds values present in both lists.

def common_elements(list1, list2):
    return list(set(list1) & set(list2))

print(common_elements([1, 2, 3], [2, 3, 4, 5]))
```

#### 6. Elements in One List but Not Another

```python
# Use: Finds elements present in first list but missing in second list.

def difference_elements(list1, list2):
    return list(set(list1) - set(list2))

print(difference_elements([1, 2, 3, 4], [2, 4]))
```
