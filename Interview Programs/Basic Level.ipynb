#### 1. Reverse a string

```python
# Method 1: Using slicing

def reverse_string(string):
    return string[::-1] # Why: [::-1] means start from end and move backward.

print(reverse_string("Hello"))
```

```python
# Method 2: Using loop

text = 'Hello'
reverse_text = ''

for char in text:
    reverse_text = char + reverse_text # Why: Each character is added before the existing string.

print(reverse_text)
```

```python
# Method 3: Using reversed()

text = 'Hello'
reverse_text = ''.join(reversed(text)) # Why: reversed() reverses characters, and join() combines them.

print(reverse_text)
```

#### 2. Find the second largest number in a list

```python
# Method 1: Using set() and sort()

def second_largest(nums):
    unique_nums = list(set(nums))

    if len(unique_nums) < 2:
        return None

    unique_nums.sort(reverse=True)

    return unique_nums[1]

print(second_largest([10, 20, 4, 45, 99, 99]))
```

```python
# Method 2: Using set() and sort()

numbers = [10, 20, 5, 30, 30]

unique_numbers = list(set(numbers)) # Why: set() removes duplicates. After sorting, second last value is second largest.
unique_numbers.sort()

print(unique_numbers[-2])
```

```python
# Method 2: Without sorting

numbers = [10, 20, 5, 30, 30]

largest = second_largest = float("-inf")

for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print(second_largest) # Key point: This is better for interviews because it avoids sorting.
```

#### 3. Find missing number from 1 to N

```python
# Method 1: Using sum formula:
# Formula: Sum of 1 to N = N * (N + 1) / 2

def find_missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(nums)

    return expected_sum - actual_sum

print(find_missing_number([1, 2, 4, 5], 5))
```

```python
# Method 2: Using set
# Key point: Easy to understand, but slower for large lists.

numbers = [1, 2, 4, 5]
n = 5

for i in range(1, n + 1):
    if i not in numbers:
        print(i)
```

#### 4. Count frequency of words in a sentence

```python
def word_frequency(sentence):
    words = sentence.lower().split()
    freq = {}

    for word in words:
        freq[word] = freq.get(word, 0) + 1

    return freq

print(word_frequency("Python is easy Python is powerful"))
```

#### 5. Remove duplicates from a list

```python
def remove_duplicates(nums):
    result = []
    seen = set()

    for num in nums:
       if num not in seen:
            result.append(num)
            seen.add(num)

    return result

print(remove_duplicates([1, 2, 2, 3, 1, 4]))
```

#### 6. Check Even or Odd

```python
def is_even_odd(num):
    if num % 2 == 0:
        return "Even"
    return "Odd"

print(is_even_odd(10))
print(is_even_odd(7))
```

#### 7. Check Prime Number

```python
def is_prime(num):
    if num <= 1:
        return False

    for i in range(2, int(num ** 0.5) + 1 + 1): # Why sqrt logic? We do not need to check all numbers up to num. Checking up to square root is enough.
        if num % i == 0:
            return False

    return True

print(is_prime(7))
print(is_prime(10))
```

#### 8. Fibonacci Series

```python
def fibonacci(n):
    a, b = 0, 1
    result = []

    for _ in range(n):
        result.append(a)
        a, b = b, a + b

    return result

print(fibonacci(7))
```

#### 9. Factorial of a Number

```python
def factorial(num):
    result = 1

    for i in range(1, num + 1):
        result *= i

    return result

print(factorial(5))
```

#### 10. Check Number Palindrome

```python
def is_number_palindrome(num):
    return str(num) == str(num)[::-1]

print(is_number_palindrome(121))
print(is_number_palindrome(123))
print(is_number_palindrome(1441))
```

#### 11. Sum of Digits

```python
def sum_of_digits(num):
    total = 0

    while num > 0:
        total += num % 10 # num % 10 gives last digit
        num //= 10 # num // 10 removes last digit

    return total

print(sum_of_digits(1234))
#print(sum_of_digits(95))
```

#### 12. Count Vowels and Consonants

```python
def count_vowels_consonants(text):
    vowels = 'aeiouAEIOU'
    vowel_count = 0
    consonants_count = 0

    for char in text:
        if char.isalpha():
            if char.lower() in vowels:
                vowel_count += 1
            else:
                consonants_count += 1

    return vowel_count, consonants_count

print(count_vowels_consonants("Python AI"))
```

#### 13. Largest Number in List

```python
def find_largest(numbers):
    largest = numbers[0]

    for num in numbers:
        if num > largest: # For smallest make: num < largest
            largest = num

    return largest

print(find_largest([10, 5, 30, 20]))
```

#### 14. Average of Numbers

```python
def calculate_average(numebrs):
    return sum(numebrs) / len(numebrs)

print(calculate_average([10, 20, 30]))
```

#### 15. Count Positive, Negative and Zero

```python
def count_numbers(numbers):
    positive = negative = zero = 0

    for num in numbers:
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        else:
            zero += 1

    return positive, negative, zero

print(count_numbers([1, -2, 0, 5, -7]))
```

#### 16. Swap Two Variables

```python
def swap_values(a, b):
    a, b = b, a
    return a, b

print(swap_values(10, 20))
```

#### 17. Check Leap Year

```python
def is_leap_year(year):
    if year % 400 == 0:
        return True
    if year % 100 == 0:
        return False
    if year % 4 == 0:
        return True
    return False

print(is_leap_year(2000))
print(is_leap_year(2026))
print(is_leap_year(1990))
```

#### 18. Celsius to Fahrenheit

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32 # Fahrenheit = (Celsius × 9/5) + 32

print(celsius_to_fahrenheit(25))
```

#### 19. Multiplication Table

```python
def multiplication(num):
    for i in range(1, 11):
        print(f'{num} x {i} = {num * i}')

print(multiplication(5))
```
