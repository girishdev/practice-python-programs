## 61 to 70: Recursion

#### Trick 67: Recursive power.

```python
def power(base, exp):
    if exp == 0:
        return 1

    return base * power(base, exp - 1)

print(power(2, 5))
```

#### Trick 68: Palindrome recursion.

```python
def is_palindrome(s):
    if len(s) <= 1:
        return True

    return s[0] == s [-1] and is_palindrome(s[1:-1])

print(is_palindrome("madam"))
```

#### Trick 69: Recursive maximum.

```python
def max_value(nums):
    if len(nums) == 1:
        return nums[0]

    return max(nums[0], max_value(nums[1:]))

print(max_value([3, 9, 1, 7]))
```

#### Trick 70: Recursive binary search.

```python
def binary_search(nums, target, left, right):
    if left > right:
        return -1

    mid = (left + right) // 2

    if nums[mid] == target:
        return mid
    elif nums[mid] < target:
        return binary_search(nums, target, mid + 1, right)
    else:
        return binary_search(nums, target, left, mid - 1)

nums = [1, 3, 5, 7, 9]
print(binary_search(nums, 7, 0, len(nums) - 1))
```

#### Trick 91: Two Sum.
**Definition:** Two Sum uses a hash map of visited values to find a target pair.

**Where used:** Matching two transaction values to a total.

**Interview question:** What is the average-case time complexity of Two Sum?

**Note:** Fixed top-level return, premature return and unreachable assignment.

```python
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []


print(two_sum([2, 7, 11, 15], 9))
```

#### Trick 92: Valid parentheses.
**Definition:** A stack tracks open brackets in last-in-first-out order.

**Where used:** Validating expressions and parsers.

**Interview question:** Why is a stack appropriate for nested parentheses?

**Note:** Fixed top-level return and rejection of unexpected characters.

```python
def valid_parentheses(s):
    stack = []
    pairs = { ")": "(", "}": "{", "]": "["}
    for ch in s:
        if ch in "({[":
            stack.append(ch)
        elif ch in pairs:
            if not stack or stack[-1] != pairs[ch]:
                return False
            stack.pop()
        else:
            return False
    return not stack


print(valid_parentheses("{[()]}"))
```

#### Trick 93: Move zeroes.

```python
nums = [0, 1, 0, 3, 12]
pos = 0

for num in nums:
    if num != 0:
        nums[pos] = num
        pos += 1

while pos < len(nums):
    nums[pos] = 0
    pos += 1

print(nums)
```

#### Trick 94: Maximum subarray using Kadane's algorithm.

```python
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
current = nums[0]
best = nums[0]

for num in nums[1:]:
    current = max(num, current + num)
    best = max(best, current)

print(best)
```

#### Trick 95: Binary search.

```python
nums = [1, 3, 5, 7, 9]
target = 7
left = 0
right = len(nums) - 1

while left <= right:
    mid = (left + right) // 2
    if nums[mid] == target:
        print("Found at index", mid)
        return

    elif nums[mid] < target:
        left = mid + 1

    else:
        right = mid - 1

print("Not found")
```

#### Trick 97: Merge two sorted arrays.

```python
a = [1, 3, 5]
b = [2, 4, 6]
i = j = 0
result = []

while i < len(a) and j < len(b):
    if a[i] < b[j]:
        result.append(a[i])
        i += 1
    else:
        result.append(b[j])
        j += 1

result.extend(a[i:])
result.extend(b[j:])

print(result)
```
