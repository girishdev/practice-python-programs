#### 1. Class and Object

```python
class Student: # Class = blueprint/template
    pass

student1 = Student() # Object = real instance of a class
print(student1)
```

#### 2. Constructor `__init__()` and `self`

```python
class Student:
    def __init__(self, name, age): # Constructor runs automatically when object is created
        self.name = name
        self.age = age

student1 = Student("Girish", 37)
print(student1.name)
print(student1.age)
```
